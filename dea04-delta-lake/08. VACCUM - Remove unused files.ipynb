### Remove unused files using VACUUM

VACCUUM is used to remove old, unused files from Delta Lake to free up storage, Permanently deletes files that are no longer referenced in the transaction log and older than retention threshold (default 7 days)

##### 1. Check the table history

```sql
DESC HISTORY demo.delta_lake.optimize_stock_prices;
```

##### 2. Run VACUUM

```sql
VACUUM demo.delta_lake.optimize_stock_prices;
```

```sql
DESC HISTORY demo.delta_lake.optimize_stock_prices;
```

```sql
ALTER TABLE demo.delta_lake.optimize_stock_prices
SET TBLPROPERTIES ('delta.deletedFileRetentionDuration' = '0 HOURS');
```

```sql
DESC HISTORY demo.delta_lake.optimize_stock_prices;
```

```sql
SELECT * FROM demo.delta_lake.optimize_stock_prices;
```

```sql
VACUUM demo.delta_lake.optimize_stock_prices RETAIN 0 HOURS;
```

```sql
SELECT * FROM demo.delta_lake.optimize_stock_prices VERSION AS OF 3;
```
