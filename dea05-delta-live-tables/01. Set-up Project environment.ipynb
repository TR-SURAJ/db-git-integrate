### Set-up Project environment for CircuitBox Data Lakehouse
1. Create external location - dea_course_ext_dl_circuitbox
2. Create Catalog - circuitbox
3. Create Schemas
-     landing
-     lakehouse
4. Create Volume - operational_data

##### 1. Create External Location
**External loction name:** dea_course_ext_dl_circuitbox\
ADLS Path: abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/\
_Storage Credential:_ dea_wrkspc_ext_sc

```sql
%fs ls "abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/"
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_circuitbox
URL 'abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL dea_wrkspc_ext_sc)
COMMENT 'External Location for the circuitbox data lakehouse';
```

```sql
%fs ls "abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/"
```

##### 2. Create catalog
**Catalog Name:** circuitbox\
_Managed Location:_ abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/

```sql
CREATE CATALOG IF NOT EXISTS circuitbox
  MANAGED LOCATION 'abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/'
  COMMENT 'Catalog for the circuitbox data lakehouse';
```

```sql
SHOW CATALOGS;
```

##### 3. Create Schemas
1. Schema name: landing\
Managed Location: abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/landing
2. Schema Name: lakehouse\
Managed Location: abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/lakehouse

```sql
USE CATALOG circuitbox;

CREATE SCHEMA IF NOT EXISTS landing
  MANAGED LOCATION 'abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/landing';

CREATE SCHEMA IF NOT EXISTS lakehouse
  MANAGED LOCATION 'abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/lakehouse';
```

```sql
SHOW SCHEMAS;
```

##### 4. Create Volume\
Volume Name: operational_data\
ADLS Path: abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/landing/operational_data

```sql
USE CATALOG circuitbox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
LOCATION 'abfss://circuitbox@deawrkspcextdl.dfs.core.windows.net/landing/operational_data';
```

```sql
%fs ls /Volumes/circuitbox/landing/operational_data
```
