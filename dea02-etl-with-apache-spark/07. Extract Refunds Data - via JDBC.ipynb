### Extract Data From the Returns SQL Table
1. Create Bronze Schema in Hive Metastore
2. Create External Table

_Note: Unity Catalog doesnt support connection to external datasources like azure sql-db, If we have to connect to external databases we have to use Lakehouse Federation. For now we are using hive_metastore which allows us to connect to external dbs

##### 1. Create Bronze Schema in Hive Metastore

```sql
CREATE SCHEMA IF NOT EXISTS hive_metastore.bronze;
```

##### 2. Create External Table

```sql
CREATE TABLE IF NOT EXISTS hive_metastore.bronze.refunds
USING JDBC
OPTIONS (
  url 'jdbc:sqlserver://adf-sql-db.database.windows.net:1433;database=adf-asql-db',
  dbtable 'Refunds',
  user 'admin123',
  password 'Suraj123'
)
```

```sql
SELECT * FROM hive_metastore.bronze.refunds;
```
